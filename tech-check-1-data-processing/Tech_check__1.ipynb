## Overview
The goal of this assignment is to demonstrate your ability to import data from various sources (Files and APIs), manipulate that data using Pandas, and visualize the results using Matplotlib.

## Step 1 : Import Necessary Libraries
    Import the standard libraries required for data manipulation, web requests, and charting. 

```python
import pandas as pd
import matplotlib.pyplot as plt
import requests
```

## Step 2 : Load Data from a Web-Based Source (API)

```python
api_url = "https://catfact.ninja/facts"
# 1. Make a get request
response = requests.get(api_url)
# 2. Parse JSON response
json_data = response.json()
# 3. Create DataFrame (The actual facts are inside the 'data' key)
df_api = pd.DataFrame(json_data['data'])
# 4. Display the results
print("--- Web-based Data (API) ---")
display(df_api.head())
```

## Step 3 : Load Data from a File-Based Source (CSV)

```python
csv_filename = "student_scores.csv"
df_csv = pd.read_csv('student_scores.csv')
print("---File-based Data (CSV) ---")
display(df_csv.head())
```

## Step 4 : Data Maipulation and Basic Calculations

```python
# 1. Extract the column
fact_lengths = df_api['length']
# 2. Calculate basic statistics
max_length = fact_lengths.max()
min_length = fact_lengths.min()
mean_length = fact_lengths.mean()
# 3. Print the results
print("\n--- Basic Calculations ---")
print(f"Maximum fact length: {max_length}")
print(f"Minimum fact length: {min_length}")
print(f"Average fact length: {mean_length:.1f}")
```

## Step 5 : Data Visualization using Matplotlib

```python
# 1. Set the figure size
plt.figure(figsize=(10,5))

# 2. Create a bar chart (X = index, Y = length of the fact)
plt.bar(df_api.index, df_api['length'], color='cornflowerblue')

# 3. Add Title and Labels
plt.title('Length of Cat Facts (API Data)')
plt.xlabel('Fact Index')
plt.ylabel('Number of characters')

# 4. Show the chart
plt.show()
```
