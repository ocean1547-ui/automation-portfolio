# Tech Check 1 - Practice: File Data Source (CSV)
Practice notebook mirroring Tech Check 1 Part A. Loads population data from a CSV file, displays basic statistics, and visualizes the top 5 countries by population in 2025.

**Data:** `practice_population_2020-2025.csv` (sample data, population in millions)

```python
# Import usual libraries
import pandas as pd
import matplotlib.pyplot as plt

# BONUS: Turn off default scientific notation and display numbers in regular format with commas
pd.options.display.float_format = '{:,.2f}'.format
```

## 1. Load Data & Display Basic Information

```python
# Load the CSV data into a Pandas DataFrame
df_a = pd.read_csv('practice_population_2020-2025.csv')

# Output basic information
print("--- DataFrame Info ---")
df_a.info()

print("\n--- DataFrame Stats ---")
display(df_a.describe())
```

## 2. Data Visualization (Top 5 Countries for Population in 2025)

```python
# Note: adjust 'Country' and '2025' if your dataset uses different column names.
# Check with df_a.head() first if you get an error.

# Sort by 2025 population in descending order and get the top 5
top_5_pop = df_a.sort_values(by='2025', ascending=False).head(5)

# Plot the Matplotlib chart
plt.figure(figsize=(10, 6))
plt.bar(top_5_pop['Country'], top_5_pop['2025'], color='steelblue')
plt.title('Top 5 Countries for Population in 2025')
plt.xlabel('Country')
plt.ylabel('Population in 2025 (millions)')
plt.show()
```

# Tech Check 1 - Practice: Web Data Source (API)
Practice notebook mirroring Tech Check 1 Part B. Extracts data from the Open Brewery DB API, displays basic stats, and visualizes the number of brewery types in Colorado.

**Data Source:** [Open Brewery DB Documentation](https://www.openbrewerydb.org/documentation)

```python
# Import usual libraries including 'requests' for the JSON API
import pandas as pd
import matplotlib.pyplot as plt
import requests
```

## 1. Load API Data & Display Basic Information

```python
# Set up the API URL for breweries in Colorado
# (per_page=200 fetches enough data)
api_url = "https://api.openbrewerydb.org/v1/breweries?by_state=colorado&per_page=200"

# Request data and load into Pandas DataFrame
response = requests.get(api_url)
df_b = pd.DataFrame(response.json())

# Output basic information
print("--- DataFrame Info ---")
df_b.info()

print("\n--- DataFrame Stats ---")
display(df_b.describe(include='all'))  # include='all' for mostly-categorical data
```

## 2. Data Visualization (Brewery Types in Colorado)

```python
# Use .value_counts() to get the count of each brewery type
brewery_type_counts = df_b['brewery_type'].value_counts()

# Plot the Matplotlib chart
plt.figure(figsize=(10, 6))
brewery_type_counts.plot(kind='bar', color='teal')

plt.title('Number of Brewery Types in Colorado')
plt.xlabel('Brewery Type')
plt.ylabel('Count')
plt.xticks(rotation=45)  # rotate labels so they don't overlap
plt.tight_layout()
plt.show()
```
