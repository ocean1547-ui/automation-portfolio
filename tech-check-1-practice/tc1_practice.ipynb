# Tech Check 1 - Practice: File Data Source (CSV)
Practice notebook mirroring the Tech Check 1 exam format. Loads product revenue data from a CSV file, displays basic statistics, and visualizes the top 5 products by revenue in 2025.

**Data:** `practice_products_2020-2025.csv` (sample data, revenue in $ millions)

```python
# Import usual libraries
import pandas as pd
import matplotlib.pyplot as plt

# BONUS: Turn off default scientific notation and display numbers in regular format with commas
pd.options.display.float_format = '{:,.2f}'.format
```

## 1. Load Data & Display Basic Information

```python
# Load the CSV data into a Pandas DataFrame
df_a = pd.read_csv('practice_products_2020-2025.csv')

# Output basic information
print("--- DataFrame Info ---")
df_a.info()

print("\n--- DataFrame Stats ---")
display(df_a.describe())
```

## 2. Data Visualization (Top 5 Products by Revenue in 2025)

```python
# Note: adjust 'Product' and '2025' if your dataset uses different column names.
# Check with df_a.head() first if you get an error.

# Sort by 2025 revenue in descending order and get the top 5
top_5_revenue = df_a.sort_values(by='2025', ascending=False).head(5)

# Plot the Matplotlib chart
plt.figure(figsize=(10, 6))
plt.bar(top_5_revenue['Product'], top_5_revenue['2025'], color='darkorange')
plt.title('Top 5 Products by Revenue in 2025')
plt.xlabel('Product')
plt.ylabel('Revenue in 2025 ($ millions)')
plt.show()
```

# Tech Check 1 - Practice: Web Data Source (API)
Practice notebook mirroring the Tech Check 1 exam format. Extracts todo items from the JSONPlaceholder API, displays basic stats, and visualizes how many todos are completed vs pending.

**Data Source:** [JSONPlaceholder](https://jsonplaceholder.typicode.com/) (free fake REST API for testing)

```python
# Import usual libraries including 'requests' for the JSON API
import pandas as pd
import matplotlib.pyplot as plt
import requests
```

## 1. Load API Data & Display Basic Information

```python
# Set up the API URL for todo items
api_url = "https://jsonplaceholder.typicode.com/todos"

# Request data and load into Pandas DataFrame
response = requests.get(api_url)
df_b = pd.DataFrame(response.json())

# Output basic information
print("--- DataFrame Info ---")
df_b.info()

print("\n--- DataFrame Stats ---")
display(df_b.describe(include='all'))  # include='all' for mixed-type data
```

## 2. Data Visualization (Todo Completion Status)

```python
# Use .value_counts() to get the count of completed vs pending todos
completed_counts = df_b['completed'].value_counts()

# Plot the Matplotlib chart
plt.figure(figsize=(10, 6))
completed_counts.plot(kind='bar', color='mediumseagreen')

plt.title('Todo Completion Status')
plt.xlabel('Completed')
plt.ylabel('Count')
plt.xticks(rotation=0)  # keep True/False labels horizontal
plt.tight_layout()
plt.show()
```
